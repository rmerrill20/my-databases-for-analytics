```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:A09a11S28!@localhost:5432/World"
)
```

```python
chart_query = """
SELECT TRIM(c.name) AS name,
       COUNT(cl.language) AS num_languages
FROM country AS c
JOIN countrylanguage AS cl
  ON c.code = cl.countrycode
WHERE cl.isofficial = 'T'
GROUP BY TRIM(c.name)
HAVING COUNT(cl.language) > 2
ORDER BY num_languages DESC;
"""

chart_df = pd.read_sql(chart_query, engine)

chart_df.plot(
    kind="bar",
    x="name",
    y="num_languages",
    figsize=(12, 6),
    legend=False
)

plt.xlabel("Country")
plt.ylabel("Number of Official Languages")
plt.title("Countries with More Than Two Official Languages")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
